```python
grades = {'Hina': 90, 'Ahmed': 85, 'Menna': 'Absent', 'Ali': 70}

def calculate_average_score(grades_dict):
    total = 0
    count = 0
    for student, score in grades_dict.items():
        try:

            numeric_score = float(score)
            total += numeric_score
            count += 1
        except ValueError:
            print(f"Skipping {student}: grade is not a valid number ({score})")

    if count == 0:
        return 0
    return total / count

average = calculate_average_score(grades)
print("Average Score:", average)
```
